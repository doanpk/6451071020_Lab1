## Câu 5: Tiền xử lý dữ liệu thiếu
**Yêu cầu:** Điền trung bình/trung vị cho cột số `Cons.price.idx`, điền mode cho cột phân loại `Marital`, xử lý `'unknown'`. Viết nhận xét bằng Markdown.

```python
print("Vi Duy Đoàn - 6451071020")
# 1. Điền giá trị trung bình cho cột số Cons.price.idx
mean_cons_price = df['Cons.price.idx'].mean()
df['Cons.price.idx'] = df['Cons.price.idx'].fillna(mean_cons_price)
print(f"Giá trị trung bình dùng để điền Cons.price.idx: {mean_cons_price:.3f}")

# 2. Điền giá trị phổ biến nhất (mode) cho cột phân loại Marital
mode_marital = df['Marital'].mode()[0]
df['Marital'] = df['Marital'].fillna(mode_marital)
print(f"Giá trị phổ biến nhất (mode) dùng để điền Marital: '{mode_marital}'")

# 3. Xử lý giá trị 'unknown' trong cột Default (thay thế bằng giá trị phổ biến nhất khác unknown)
mode_default = df['Default'][df['Default'] != 'unknown'].mode()[0]
df['Default'] = df['Default'].replace('unknown', mode_default)
print(f"Giá trị thay thế cho 'unknown' trong cột Default: '{mode_default}'")

# 4. Kiểm tra lại toàn bộ giá trị khuyết sau xử lý
print("\n=== SỐ LƯỢNG GIÁ TRỊ RỖNG CÒN LẠI SAU TIỀN XỬ LÝ ===")
print(df.isnull().sum())
print(f"Số lượng giá trị 'unknown' còn lại trong df: {(df == 'unknown').sum().sum()}")

print("\nDữ liệu sau khi xử lý thiếu:")
display(df.head(10))
```

## Câu 6: Mã hóa thuộc tính phân loại và Rời rạc hóa cột Duration
**Yêu cầu:** Mã hóa các thuộc tính phân loại (`Job`, `Marital`, `Education`, `Default`, `Loan`, `Subscribe`) sang dạng số và rời rạc hóa cột `Duration` thành các nhóm. Hiển thị kết quả cuối cùng.

```python
print("Vi Duy Đoàn - 6451071020")
# 1. Rời rạc hóa cột Duration thành 3 nhóm: 
# Nhóm 0: Ngắn (<= 300s), Nhóm 1: Trung bình (300 - 800s), Nhóm 2: Dài (> 800s)
duration_bins = [-1, 300, 800, 3000]
duration_labels = [0, 1, 2]
df['Duration_Group'] = pd.cut(df['Duration'], bins=duration_bins, labels=duration_labels).astype(int)

# 2. Mã hóa các thuộc tính phân loại sang dạng số (Categorical Encoding)
categorical_cols = ['Job', 'Marital', 'Education', 'Default', 'Loan', 'Subscribe']

# Tạo từ điển ánh xạ rõ ràng để dễ giải thích mô hình
encoding_dict = {}
for col in categorical_cols:
    df[col] = df[col].astype('category')
    encoding_dict[col] = dict(enumerate(df[col].cat.categories))
    df[col] = df[col].cat.codes

print("=== BẢNG ÁNH XẠ MÃ HÓA CÁC THUỘC TÍNH (ENCODING MAPPING) ===")
for col, mapping in encoding_dict.items():
    print(f"- {col}: {mapping}")

# 3. Hiển thị toàn bộ bảng dữ liệu sau cùng
print("\n=== TOÀN BỘ 16 DÒNG DỮ LIỆU HOÀN THIỆN SAU TIỀN XỬ LÝ VÀ MÃ HÓA ===")
display(df)

# 4. Kiểm tra cấu trúc dữ liệu tổng quát lần cuối
print("\n=== THÔNG TIN BỘ DỮ LIỆU HOÀN THÀNH (df.info()) ===")
df.info()
```

### Nhận xét Câu 6:
1. **Hiệu quả của việc Rời rạc hóa `Duration`:**
   - Việc chia `Duration` thành 3 thang đo: `0` (ngắn), `1` (trung bình), `2` (dài) giúp mô hình máy học dễ dàng phân biệt các mức độ quan tâm của khách hàng mà không bị ảnh hưởng bởi sự chênh lệch giây đo lường quá chi tiết.
2. **Mã hóa số học (Numeric Encoding):**
   - Toàn bộ các thuộc tính chữ (`Job`, `Marital`, `Education`, `Default`, `Loan`, `Subscribe`) đều đã được chuyển đổi thành các mã số nguyên đại diện, bảo toàn toàn vẹn thông tin phân loại.
3. **Tổng kết bài toán Bank Marketing:**
   - Bộ dữ liệu mẫu đã hoàn tất đầy đủ quy trình tiền xử lý chuẩn khoa học dữ liệu: làm sạch lỗi định dạng dấu phẩy số thực, phát hiện và xử lý giá trị khuyết / giá trị `unknown`, trực quan hóa khám phá tri thức (EDA) và mã hóa đặc trưng hoàn chỉnh.
   - Toàn bộ 16 mẫu hiện tại đều ở dạng dữ liệu số (`int64`, `int8`, `float64`), sẵn sàng để đưa vào áp dụng các thuật toán khai phá quy luật như Cây quyết định (Decision Tree) hoặc Luật kết hợp (Association Rules).
